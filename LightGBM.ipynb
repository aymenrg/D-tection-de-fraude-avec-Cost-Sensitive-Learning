```python
for column in df.columns:
  most_common_value = df[column].mode()[0]
  df[column] = df[column].fillna(most_common_value)
```

```python
df.drop_duplicates(inplace = True)
```

```python
y = df['isFraud']
df = df.drop('isFraud', axis=1)
df = df.drop(columns=['Num']+columns_to_drop)

# Selecting numerical and categorical data
num_df = df.select_dtypes(include=['int64', 'float64'])
cat_df = df.select_dtypes(include=['object', 'category'])

# Initialize a label encoder for each categorical column in cat_df
for column in cat_df.columns:
    le = LabelEncoder()
    cat_df[column] = le.fit_transform(cat_df[column])

# Concatenate the numerical data and the newly encoded categorical data
final_df = pd.concat([num_df.reset_index(drop=True), cat_df.reset_index(drop=True)], axis=1)
```

```python
model_lgb = LGBMClassifier(n_estimators=400,
                          learning_rate=0.05,
                          num_leaves=31,
                          colsample_bytree=0.8,
                          subsample_freq=3,
                          subsample=0.8,
                          random_state=42)

model_lgb.fit(X_train, y_train)

y_pred_lgb = model_lgb.predict(X_test)
y_pred_lgb_proba = model_lgb.predict_proba(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_lgb))
print("Precision:", precision_score(y_test, y_pred_lgb))
print("Recall:", recall_score(y_test, y_pred_lgb))
print("F1 Score:", f1_score(y_test, y_pred_lgb))
print("ROC AUC Score:", roc_auc_score(y_test, y_pred_lgb_proba[:, 1]))
```

```python
cm = confusion_matrix(y_test, y_pred_lgb)
print("Confusion Matrix:")
print(cm)
```

```python
param_grid = {
    'learning_rate': [0.01, 0.05, 0.1],
    'num_leaves': [31, 63, 127],
    'colsample_bytree': [0.6, 0.7, 0.8],
    'subsample_freq': [1, 2, 3],
    'subsample': [0.6, 0.7, 0.8],
}

grid_search = GridSearchCV(model_lgb, param_grid, cv=5, n_jobs=-1, verbose=1)
grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)

best_model = grid_search.best_estimator_

y_pred_best = best_model.predict(X_test)
y_pred_proba_best = best_model.predict_proba(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_best))
print("Precision:", precision_score(y_test, y_pred_best))
print("Recall:", recall_score(y_test, y_pred_best))
print("F1 Score:", f1_score(y_test, y_pred_best))
print("ROC AUC Score:", roc_auc_score(y_test, y_pred_proba_best[:, 1]))

cm = confusion_matrix(y_test, y_pred_best)
print("Confusion Matrix:")
print(cm)
```
